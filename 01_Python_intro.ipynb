## 2. Variáveis, tipos e operadores
---

Uma **variável** é um nome associado a um valor. Não é preciso declarar o tipo:
o Python o descobre sozinho.

| Tipo | Exemplo | Uso típico aqui |
|---|---|---|
| `int` | `21` | número de vértices, quantidades inteiras |
| `float` | `2665.0` | custos, distâncias, tempos |
| `str` | `"Niterói"` | rótulos de vértices e atividades |
| `bool` | `True` / `False` | o grafo é conexo? a solução é viável? |

````python
n_cidades = 21           # int
distancia = 95.5         # float
origem    = "Niterói"    # str
conexo    = True         # bool

print(n_cidades, type(n_cidades))
print(distancia, type(distancia))
print(origem,    type(origem))
print(conexo,    type(conexo))
````

**Operadores aritméticos e de comparação.** Atenção a dois pontos que costumam
gerar erro: `=` é **atribuição** e `==` é **comparação**; e `/` sempre devolve
`float`, enquanto `//` faz a divisão inteira.

````python
a, b = 17, 5              # atribuição múltipla em uma linha

print("soma .............:", a + b)
print("divisão ..........:", a / b)     # 3.4  (float)
print("divisão inteira ..:", a // b)    # 3
print("resto ............:", a % b)     # 2
print("potência .........:", a ** 2)
print()
print("a == b ?", a == b)               # comparação
print("a >  b ?", a >  b)
print("a != b e b > 0 ?", a != b and b > 0)
````

## 3. f-strings: montando relatórios legíveis
---

Praticamente toda aula termina com um relatório do tipo *"a solução ótima custa
X e usa os trechos Y"*. A forma mais prática de montar esse texto é a **f-string**:
coloque um `f` antes das aspas e escreva as variáveis entre chaves `{}`.

Dentro das chaves cabe qualquer expressão, e depois de `:` vai a **formatação**:

| Formato | Efeito |
|---|---|
| `{x:.2f}` | duas casas decimais |
| `{x:.1%}` | porcentagem com uma casa |
| `{x:>8}` | alinhado à direita em 8 colunas |
| `{x:<12}` | alinhado à esquerda em 12 colunas |

````python
cidade   = "Niterói"
custo    = 2665.0
economia = 0.5554

print(f"Solução para {cidade}: custo total de R$ {custo:.2f}")
print(f"Economia obtida: {economia:.1%}")
print(f"Custo por cidade: {custo / 21:.1f}")
print()

# alinhamento — muito útil para imprimir tabelas simples
for nome, km in [("Niterói", 15), ("Petrópolis", 70), ("Nova Friburgo", 160)]:
    print(f"{nome:<15} {km:>5} km")
````

## 5. Tuplas e desempacotamento
---

Uma **tupla** é uma sequência **ordenada** e **imutável** — escrita com parênteses.

Este é, provavelmente, o tópico mais importante desta revisão: **no `networkx`,
uma aresta é uma tupla**. A rede inteira costuma ser descrita como uma lista de
tuplas `(origem, destino, peso)`.

````python
aresta  = ("Niterói", "Maricá", 40)          # (origem, destino, distância)
arestas = [("Rio", "Niterói",  15),
           ("Niterói", "Maricá", 40),
           ("Maricá", "Cabo Frio", 110)]

print("uma aresta :", aresta)
print("o peso dela:", aresta[2])
print("a rede tem", len(arestas), "trechos")
````

**Desempacotamento**: podemos atribuir os elementos de uma tupla a várias
variáveis de uma só vez. É assim que percorreremos as arestas de um grafo o tempo
todo — repare no `for u, v, peso in arestas`.

````python
u, v, peso = aresta                  # desempacotando
print(f"de {u} para {v}, {peso} km")
print()

total = 0
for u, v, peso in arestas:           # desempacotando a cada volta do laço
    print(f"{u:<10} -> {v:<12} {peso:>4} km")
    total = total + peso

print(f"\nDistância total: {total} km")
````

## 6. Dicionários
---

Um **dicionário** associa **chaves** a **valores**. É a estrutura por trás de quase
tudo no `networkx` (atributos de vértices e arestas) e no `pyomo` (parâmetros
indexados do modelo).

````python
distancias = {"Rio": 0, "Niterói": 15, "Maricá": 55, "Cabo Frio": 165}

print("valor associado a 'Niterói':", distancias["Niterói"])
print("chaves  :", list(distancias.keys()))
print("valores :", list(distancias.values()))
print("'Macaé' está no dicionário?", "Macaé" in distancias)

distancias["Macaé"] = 255            # acrescenta um novo par
del distancias["Rio"]                # remove um par
print("\ndepois das alterações:", distancias)
````

Percorrer um dicionário com `.items()` devolve, a cada volta, a chave **e** o
valor — outro desempacotamento de tupla.

````python
for cidade, km in distancias.items():
    print(f"{cidade:<12} {km:>4} km")

print(f"\nMaior distância: {max(distancias.values())} km")
````

> **Chaves podem ser tuplas.** Isso é o que permite representar parâmetros com
> dois índices, como o custo $c_{ij}$ de ir de $i$ até $j$ — exatamente a forma
> que usaremos nos modelos do Pyomo.

````python
custo = {("Rio", "Niterói"): 15,
         ("Niterói", "Maricá"): 40,
         ("Maricá", "Cabo Frio"): 110}

print("custo de Niterói a Maricá:", custo[("Niterói", "Maricá")])
````

## 7. Conjuntos
---

Um **conjunto** (`set`) guarda elementos **sem repetição** e **sem ordem**. Usaremos
conjuntos principalmente para marcar "vértices já visitados" e para comparar
componentes de uma rede.

````python
visitados     = {"Rio", "Niterói"}
todos         = {"Rio", "Niterói", "Maricá", "Cabo Frio"}

visitados.add("Maricá")

print("visitados .......:", visitados)
print("faltam visitar ..:", todos - visitados)      # diferença
print("em comum ........:", todos & visitados)      # interseção
print("todos foram visitados?", todos == visitados)

# uso frequente: eliminar duplicatas de uma lista
print("\nsem duplicatas:", set([1, 2, 2, 3, 3, 3]))
````

## 9. Repetição: `for` e `while`
---

O `for` percorre qualquer sequência (lista, tupla, dicionário, conjunto, ou um
`range`). O `range(a, b, passo)` gera os inteiros de `a` até `b - 1`.

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(1, 11, 2):        # de 1 a 10, de 2 em 2
    print(i, end=" ")
print()
````

Dois auxiliares que aparecem bastante:

- `enumerate(lista)` — devolve `(posição, elemento)`;
- `zip(a, b)` — percorre duas sequências em paralelo.

````python
cidades = ["Niterói", "Maricá", "Cabo Frio"]
populacao = [516_000, 197_000, 232_000]

for i, cidade in enumerate(cidades, start=1):
    print(f"{i}º - {cidade}")

print()
for cidade, hab in zip(cidades, populacao):
    print(f"{cidade:<12} {hab:>8,} habitantes")
````

O `while` repete **enquanto** uma condição for verdadeira. Usaremos esse padrão
nas aulas de simulação, em que não sabemos de antemão quantas iterações serão
necessárias.

````python
estoque = 1000
demanda = 300
periodo = 0

while estoque >= demanda:
    estoque = estoque - demanda
    periodo = periodo + 1
    print(f"período {periodo}: sobraram {estoque} unidades")

print(f"\nApós {periodo} períodos o estoque ({estoque}) não atende mais à demanda.")
````

## 10. Compreensões e `sum(...)`
---

Uma **compreensão de lista** constrói uma lista em uma única linha. Compare:

```python
# forma longa                    # compreensão
dobros = []                      dobros = [2 * x for x in valores]
for x in valores:
    dobros.append(2 * x)
```

Isso não é só elegância: **é a forma como escreveremos funções objetivo e
restrições no Pyomo**, do tipo $\sum_{(i,j) \in A} c_{ij}\, x_{ij}$.

````python
valores = [10, 25, 4, 78, 33]

dobros    = [2 * x for x in valores]                 # transformação
grandes   = [x for x in valores if x > 20]           # filtro
rotulados = [f"v{i}" for i in range(1, 6)]           # geração

print("dobros  :", dobros)
print("grandes :", grandes)
print("rótulos :", rotulados)
````

Também existem **compreensões de dicionário** e de conjunto, com a mesma sintaxe
usando `{}`.

````python
arestas = [("A", "B", 4), ("A", "C", 2), ("B", "C", 1), ("B", "D", 5)]

pesos    = {(u, v): p for u, v, p in arestas}        # dicionário
vertices = {u for u, v, p in arestas} | {v for u, v, p in arestas}   # conjunto

print("pesos   :", pesos)
print("vértices:", sorted(vertices))
````

E o padrão mais importante de todos: **somatório com filtro**, escrito como um
`sum(...)` com uma compreensão dentro.

````python
# Custo total de todas as arestas
print("custo total       :", sum(p for u, v, p in arestas))

# Custo apenas das arestas que saem de "A"  →  soma com condição
print("custo saindo de A :", sum(p for u, v, p in arestas if u == "A"))

# Contando quantas satisfazem uma condição
print("arestas baratas   :", sum(1 for u, v, p in arestas if p <= 2))
````

## 11. Ordenação com `sorted` e `lambda`
---

`sorted(sequencia)` devolve uma nova sequência ordenada. Para ordenar por um
critério específico usamos o parâmetro `key`, que recebe uma função dizendo
**por qual valor** ordenar.

Uma função `lambda` é uma função de uma linha, sem nome — perfeita para isso.
O algoritmo de Kruskal (Aula 03), por exemplo, começa ordenando as arestas por custo.

````python
arestas = [("A", "B", 4), ("A", "C", 2), ("B", "C", 1), ("B", "D", 5)]

por_custo = sorted(arestas, key=lambda a: a[2])              # a[2] é o peso
decrescente = sorted(arestas, key=lambda a: a[2], reverse=True)

print("ordenadas por custo crescente:")
for u, v, p in por_custo:
    print(f"  {u}-{v}: {p}")

print("\na mais cara:", decrescente[0])
````

## 12. Funções
---

Uma **função** agrupa um trecho de código sob um nome, para ser reutilizado.
Definimos com `def`, e `return` devolve o resultado.

Argumentos podem ter **valor padrão**, e a chamada pode nomear os argumentos —
padrão que você verá em quase toda função do `networkx`
(ex.: `nx.minimum_spanning_tree(G, weight="distancia", algorithm="prim")`).

````python
def custo_total(arestas, desconto=0.0):
    """Soma os pesos das arestas, aplicando um desconto opcional."""
    bruto = sum(p for u, v, p in arestas)
    return bruto * (1 - desconto)


print("sem desconto     :", custo_total(arestas))
print("com 10% (posição):", custo_total(arestas, 0.10))
print("com 10% (nomeado):", custo_total(arestas, desconto=0.10))
````

## 14. Exemplo integrador
---

Para fechar, um exemplo que usa **quase tudo** que revisamos e antecipa a Aula 02:
a partir de uma rede descrita como lista de tuplas, vamos calcular quantas ligações
cada cidade tem e apresentar o resultado.

````python
rede = [
    ("Rio de Janeiro", "Niterói",        15),
    ("Rio de Janeiro", "Petrópolis",     70),
    ("Niterói",        "Maricá",         40),
    ("Niterói",        "Itaboraí",       40),
    ("Maricá",         "Cabo Frio",     110),
    ("Itaboraí",       "Nova Friburgo", 100),
    ("Petrópolis",     "Nova Friburgo",  90),
    ("Cabo Frio",      "Macaé",          90),
]

# 1) conjunto de cidades (sem repetição), via compreensão
cidades = sorted({u for u, v, km in rede} | {v for u, v, km in rede})

# 2) dicionário: quantas ligações e quantos km incidem em cada cidade
ligacoes = {c: 0 for c in cidades}
km_total = {c: 0 for c in cidades}

for u, v, km in rede:
    ligacoes[u] += 1
    ligacoes[v] += 1
    km_total[u] += km
    km_total[v] += km

# 3) apresentando em uma tabela
resumo = pd.DataFrame({"ligações": ligacoes, "km incidentes": km_total})
resumo.index.name = "cidade"
resumo = resumo.sort_values("ligações", ascending=False)

print(f"A rede tem {len(cidades)} cidades e {len(rede)} trechos, "
      f"totalizando {sum(km for u, v, km in rede)} km.\n")
resumo
````

````python
# 4) o trecho mais longo e o mais curto
mais_longo = max(rede, key=lambda t: t[2])
mais_curto = min(rede, key=lambda t: t[2])

print(f"Trecho mais longo: {mais_longo[0]} - {mais_longo[1]} ({mais_longo[2]} km)")
print(f"Trecho mais curto: {mais_curto[0]} - {mais_curto[1]} ({mais_curto[2]} km)")

# 5) um gráfico do resultado
fig, eixo = plt.subplots(figsize=(7, 4))
eixo.barh(resumo.index, resumo["ligações"], color="seagreen")
eixo.set_xlabel("número de ligações")
eixo.set_title("Quantas ligações diretas cada cidade tem")
eixo.invert_yaxis()
plt.tight_layout()
plt.show()
````
